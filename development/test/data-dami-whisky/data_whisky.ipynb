```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# 商品データ（商品ID、商品名、単価）
products = [
    (25, "余市　シングルモルト", 14520),
    (26, "宮城峡　シングルモルト", 14520),
    (27, "岡山　シングルモルト", 14680),
    (28, "小正嘉之介MBR シングルモルト", 11000),
    (29, "小正嘉之介MDS シングルモルト", 13000),
    (30, "小正嘉之介W-Ds MBR シングルモルト", 15000),
    (31, "碧AO ブレンデットウイスキー", 5640),
    (32, "サントリーローヤル ブレンデットウイスキー", 7000),
    (33, "くら ブレンデットウイスキー", 4000),
    (34, "くら ラムスカスクフィニッシュ ブレンデットウイスキー", 4500),
    (35, "倉吉 シングルモルト", 6000),
    (36, "倉吉12年 ブレンデットウイスキー", 15000),
    (37, "倉吉18年 ブレンデットウイスキー", 35000),
    (38, "倉吉25年 ブレンデットウイスキー", 100000),
    (39, "倉吉29年 ブレンデットウイスキー", 150000),
    (40, "倉吉31年 ブレンデットウイスキー", 200000),
    (41, "桜尾 シングルモルト", 6000),
    (42, "桜尾 SHERRY-CASK シングルモルト", 11950),
    (43, "桜尾四季 シングルモルト", 15000),
    (44, "あかし7年 シングルモルト", 16000),
    (45, "あかし シングルモルト", 1658),
    (46, "戸河内 ブレンデットウイスキー", 3000),
    (47, "戸河内四季 ブレンデットウイスキー", 6000),
    (48, "富士山 ブレンデットウイスキー", 2220),
    (49, "越百 ブレンデットウイスキー", 4100),
    (50, "山桜黒ラベル ブレンデットウイスキー", 1800),
    (51, "BORE MALT 山桜 ピュアウイスキー", 5500)
]


# 各商品IDに対して、1日に販売される本数の範囲（min, max）
sales_ranges = {
    25: (4, 13),
    26: (3, 12),
    27: (2, 7),
    28: (4, 8),
    29: (2, 6),
    30: (1, 5),
    31: (5, 7),
    32: (6, 10),
    33: (5, 9),
    34: (4, 11),
    35: (4, 5),  # 2日に1本売れる想定
    36: (3, 5),
    37: (2, 5),
    38: (1, 3),
    39: (1, 3),
    40: (0, 1),
    41: (5, 14),
    42: (3, 7),
    43: (4, 6),
    44: (0, 2),
    45: (7, 14),
    46: (5, 14),
    47: (0, 5),
    48: (0, 6)
}

# ダミーデータの生成対象期間（4月1日〜4月30日）
start_date = datetime(2025, 4, 1)
end_date = datetime(2025, 4, 30)
date_range = pd.date_range(start=start_date, end=end_date)

# データを保存するリスト
data = []

# 購入IDのカウンター（1から始めて連番を振る）
buy_check_id = 1

# 顧客IDは1000〜1099の100人からランダムに選ぶ
customer_id_pool = list(range(1000, 1100))

# 日ごとに全商品について販売データを生成
for date in date_range:
    for product_id, name, price in products:
        # 商品ごとの日別販売本数を乱数で決定
        min_q, max_q = sales_ranges.get(product_id, (0, 0))
        sales_count = np.random.randint(min_q, max_q + 1)

        # sales_count 回だけその商品を販売したとして記録する
        for _ in range(sales_count):
            quantity = np.random.randint(1, 4)  # 一度に買う本数（1〜3本）
            total_price = quantity * price     # 総価格 = 単価 × 本数
            customer_id = random.choice(customer_id_pool)  # ランダムな顧客ID

            # 購入時間（10時〜20時の間でランダムな時刻）
            purchase_time = date + timedelta(hours=random.randint(10, 20), minutes=random.randint(0, 59))

            # 一行分の販売データを追加
            data.append([
                buy_check_id,
                product_id,
                name,
                price,
                quantity,
                total_price,
                customer_id,
                purchase_time.strftime('%Y-%m-%d %H:%M:%S')
            ])
            buy_check_id += 1  # 購入IDをインクリメント

# pandas DataFrame に変換（列名も日本語で指定）
df = pd.DataFrame(data, columns=[
    "購入ID", "商品ID", "商品名", "単価", "数量", "総価格", "顧客ID", "購入日時"
])

# CSVファイルとして保存（index=False で行番号を出力しない）
csv_path = "/Users/yugo_suzuki/mydevelopment/development/test/data-dami-whisky/buy_check_dummy_data_whisky.csv"
df.to_csv(csv_path, index=False)

# 結果表示
print(f"CSVファイルを保存しました: {csv_path}")
```
